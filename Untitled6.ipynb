```python
# Cell 3: Clean mock SDN controller for Colab

mock_sdn = """
import time
from collections import defaultdict

class QoSAwareHandoverController:
    \"""
    Simple mock controller for Colab: provides the same API as the real SDN controller,
    but without requiring Ryu or OpenFlow.
    \"""
    def __init__(self):
        self.handover_cache = {}
        self.mn_locations = {}
        self.qos_metrics = defaultdict(dict)
        self.logger = type('L', (), {'info': print, 'warning': print})()

    def proactive_handover_preparation(self, mn_id, current_ap, target_ap):
        handover_start_time = time.time()
        self.handover_cache[mn_id] = {
            'old_ap': current_ap,
            'new_ap': target_ap,
            'timestamp': handover_start_time,
            'status': 'prepared'
        }
        self.logger.info(f"[MOCK SDN] Preparing handover for {mn_id} from {current_ap} to {target_ap}")
        return handover_start_time

    def execute_fast_handover(self, mn_id):
        if mn_id in self.handover_cache:
            ctx = self.handover_cache[mn_id]
            delay = max(0.001, time.time() - ctx['timestamp'])
            self.mn_locations[mn_id] = ctx['new_ap']
            self.logger.info(f"[MOCK SDN] Executed fast handover for {mn_id} in {delay:.4f}s")
            return delay
        return 0.0

    def calculate_shortest_path(self, src, dst):
        return [src, dst]
"""

with open("sdn_controller.py", "w") as f:
    f.write(mock_sdn)

print("Mock sdn_controller.py created successfully.")
```

```python
# Cell 5: Import your modules and run the simulation
# If you uploaded your original files, they will be imported. If not, the mock sdn_controller.py will be used.

import logging
logging.basicConfig(level=logging.INFO)

# If files were uploaded, they will be in the working dir. Use them directly.
# If you used the mock SDN controller above, the import will pick that up.

from simulation_framework import HandoverSimulation
from integrated_handover_manager import IntegratedHandoverManager
from mipv6_manager import MobileIPv6Manager
from handover_decision_engine import HybridHandoverDecision
# sdn_controller will be either your uploaded Ryu controller (if present) or the mock one created earlier
from sdn_controller import QoSAwareHandoverController

# Instantiate components
sdn_controller = QoSAwareHandoverController()
mipv6_manager = MobileIPv6Manager(home_address="2001:db8:1::1", home_agent_address="2001:db8:1::100")

topsis_weights = {
    'rssi': 0.30,
    'bandwidth': 0.25,
    'latency': 0.20,
    'packet_loss': 0.15,
    'cost': 0.10
}

decision_engine = HybridHandoverDecision(topsis_weights=topsis_weights, fuzzy_weight=0.4, topsis_weight=0.6)

handover_manager = IntegratedHandoverManager(sdn_controller, mipv6_manager, decision_engine)

# Run simulation
simulation = HandoverSimulation(handover_manager)
simulation.run_mobility_scenario(num_handovers=20)  # reduce or increase as you like
print("Simulation run finished.")
```
